```sql
-- Encuentraw el segundo salario más salto
SELECT DISTINCT salario AS segundo_salario
FROM empleados
ORDER BY salario DESC
LIMIT 1 OFFSET 1;
```

## Explicación

La idea es quedarse con los salarios únicos ordenados y saltar el más salto:

1. **DISTINCT** elimina los salarios duplicados (Ana y María ganan lo mismo).
2. **ORDER BY** salario DESC los ordena de mayor a menor.
3. **LIMIT 1 OFFSET 1** salta el primero (el más alto) y devuelve el segundo.
