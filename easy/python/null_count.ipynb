# Contar valores nulos por columna | PYTHON | fácil

Escribe una función contar_nulos(df) que reciba un DataFrame y devuelva un diccionario con el nombre de cada columna que tenga al menos un valor nulo y la cantidad de nulos que tiene. Las columnas sin nulos no aparecen en el resultado.

Ejemplo (Caso 1)

| nombre | edad | ciudad   |
| ------ | ---- | -------- |
| Ana    | 25   | Madrid   |
| Luis   | NaN  | NaN      |
| Marta  | 30   | Valencia |
| Pedro  | NaN  | Madrid   |

Resultado: {'edad': 2, 'ciudad': 1}

Output esperado
```python
{'edad': 2, 'ciudad': 1}
```


````python
import numpy as np
import pandas as pd
````

````python
df = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Marta", "Pedro"],
    "edad": [25, np.nan, 30, np.nan],
    "ciudad": ["Madrid", np.nan, "Valencia", "Madrid"],
})

df
````

````python
def null_count(df: pd.DataFrame):
    null_count = df.isna().sum()
    null_count = null_count[null_count > 0].to_dict()

    return null_count

# optimal solution
def contar_nulos(df: pd.DataFrame):
    nulos = df.isnull().sum()
    return {k: int(v) for k, v in nulos.items() if v > 0}
````

````python
df_null_count = null_count(df)

df_null_count
````
