## Enunciado

Escribe una función explotar_tags(df) que reciba un DataFrame con columnas producto (str) y tags (lista de strings), y devuelva un DataFrame donde cada tag tenga su propia fila. Ordena el resultado por producto y luego por tag, ambos alfabéticamente. Reinicia el índice.



**Ejemplo (Caso 1)**

| producto | tags |
| -------- | ---- |
| Laptop   | ["tech", "work"] |
| Libro    | ["edu"]  |

**Resultado:**

| producto | tags |
| -------- | ---- |
| Laptop   | tech |
| Laptop   | work |
| Libro    | edu  |

## Python

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'producto': ["Laptop", "Libro"],
    'tags' : [["tech",  "work"], ["edu"]]
})

df
```

```python
def explode_df(df: pd.DataFrame):
    df = (df.explode('tags', ignore_index=True)
          .sort_values(by=['producto', 'tags']))

    return df
```

```python
df = explode_df(df)

df
```

## Explicación



1. explode('tags') toma cada lista de la columna tags y la convierte en filas individuales, replicando la fila del producto.
2. sort_values(['producto', 'tags']) ordena primero por producto y luego por tag.
3. reset_index(drop=True) limpia el índice para que sea secuencial.

Es la operación inversa a un groupby + agg(list).
