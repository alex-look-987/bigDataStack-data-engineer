# Normalizar texto con str.lower/upper/title PYTHON fácil


Escribe una función normalizar_nombres(df) que reciba un DataFrame con una columna nombre (str) con nombres escritos de forma inconsistente (mayúsculas, minúsculas, mezclas) y devuelva el mismo DataFrame con tres nuevas columnas:

- **nombre_lower:** todo en minúsculas
- **nombre_upper:** todo en mayúsculas
- **nombre_title:** formato título (primera letra de cada palabra en mayúscula)

El DataFrame original no debe modificarse. Devuelve el nuevo DataFrame ordenado por la columna nombre original alfabéticamente, con el índice reiniciado.

Ejemplo (Caso 1)

| nombre         |
| -------------- |
| JUAN GARCÍA    |
| ana lópez      |
| Pedro Martínez |

Resultado:

| nombre         | nombre_lower   | nombre_upper   | nombre_title   |
| -------------- | -------------- | -------------- | -------------- |
| ana lópez      | ana lópez      | ANA LÓPEZ      | Ana López      |
| JUAN GARCÍA    | juan garcía    | JUAN GARCÍA    | Juan García    |
| Pedro Martínez | pedro martínez | PEDRO MARTÍNEZ | Pedro Martínez |

Output esperado

```python
[('JUAN GARCÍA', 'juan garcía', 'JUAN GARCÍA', 'Juan García'), ('Pedro Martínez', 'pedro martínez', 'PEDRO MARTÍNEZ', 'Pedro Martínez'), ('ana lópez', 'ana lópez', 'ANA LÓPEZ', 'Ana López')]
```

````python
import pandas as pd
````

````python
df = pd.DataFrame(
    {'nombre': [
        'JUAN GARCÍA',
        'ana lópez',
        'Pedro Martínez'
    ]}
)

df
````

````python
def names_normalization(df: pd.DataFrame):

    result = df.copy()
    
    result['nombre_lower'] = df['nombre'].str.lower()
    result['nombre_upper'] = df['nombre'].str.upper()
    result['nombre_title'] = df['nombre'].str.title()

    result = result.sort_values('nombre').reset_index(drop=True)

    return result
````

````python
normalized_names = names_normalization(df)

normalized_names
````
