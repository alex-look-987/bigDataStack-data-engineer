```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "nombre": ['Ana', 'Luis', 'Bea'],
    "edad": [25, 30, 25]
})

df
```

```python
def count_unique(df:pd.DataFrame):
    return dict(sorted(df.nunique().items()))
```

```python
df = count_unique(df)

df
```

## Explicación



1. df.nunique() devuelve una Serie con el número de valores únicos por columna.
2. .items() convierte a pares (columna, conteo).
3. sorted() ordena por nombre de columna (primer elemento del par).
4. dict() convierte de vuelta a diccionario.

Es una sola línea gracias a nunique(), que es el método nativo de pandas para esto.
