# Limitar valores con clip | PYTHON | fácil | 4/10/2026

Tienes un DataFrame con columnas: alumno (str), nota (float).

Escribe una función limitar_notas(df) que devuelva el DataFrame con las notas limitadas al rango [0, 10]:

- Si la nota es menor que 0, se convierte en 0.0.
- Si la nota es mayor que 10, se convierte en 10.0.
- Las demás quedan igual.

Devuelve el DataFrame ordenado por alumno, con el índice reseteado.


**Ejemplo (Caso 1)**

| alumno | nota |
| ------ | ---- |
| Ana    | 12.0 |
| Luis   | -3.0 |
| María  | 7.5  |

**Resultado:**

| alumno	nota |
| ----------- |
| Ana	10.0    |
| Luis	0.0    |
| María	7.5   |

**Output esperado**

```python
[('Ana', 10.0), ('Luis', 0.0), ('María', 7.5)]
```


````python
import pandas as pd
````

````python
df = pd.DataFrame({
    "alumno": ['Ana', 'Luis', 'María'],
    "nota": [12, -3, 7.5]
})

df
````

````python
def clip_values(df: pd.DataFrame):
    df = df.copy()

    df['nota'] = df['nota'].clip(lower=0, upper=10)

    df = df.sort_values(by=['alumno'], ignore_index=True)

    return df
````

````python
df = clip_values(df)

df
````

## Explicación

1. clip(lower=0, upper=10) limita los valores: por debajo de 0 se convierte en 0, por encima de 10 en 10.
2. Ordenamos por alumno y reseteamos el índice.

Es una operación muy común en preprocesamiento de datos para acotar valores a un rango válido.
