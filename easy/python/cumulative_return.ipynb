# Rendimiento acumulado con cumprod | PYTHON | fácil | 12/11/2026.

Tienes un DataFrame con columnas mes (str) y rendimiento (float), donde rendimiento es el factor multiplicativo mensual (ej: 1.05 = +5%, 0.97 = -3%).

Escribe una función rendimiento_acumulado(df) que devuelva el mismo DataFrame con una columna adicional acumulado que contenga el rendimiento acumulado (producto de todos los factores hasta ese mes inclusive).

**Ejemplo (Caso 1)**

| mes	rendimiento |
| --------------- |
| enero	1.10      |
| febrero	0.95    |
| marzo	1.05      |

**Resultado:**

| mes     | rendimiento | acumulado |
| ------- | ----------- | --------- |
| enero   | 1.10        | 1.1       |
| febrero | 0.95        | 1.045     |
| marzo   | 1.05        | 1.09725   |

**Output esperado**
```python
[('enero', 1.1), ('febrero', 1.045), ('marzo', 1.09725)]
```

````python
import pandas as pd
````

````python
df = pd.DataFrame({
    "mes": ['enero', 'febrero', 'marzo'],
    "rendimiento": [1.10, 0.95, 1.05]
})

df
````

````python
def cumulative_return(df: pd.DataFrame):
    df = df.copy()

    df['acumulado'] = df['rendimiento'].cumprod()#.round(3)

    return df
````

````python
df = cumulative_return(df)

df
````

## Explicación

1. Copiamos el DataFrame para no modificar el original.
2. cumprod() calcula el producto acumulado: cada fila es el producto de todos los valores anteriores incluyéndose a sí misma.
3. Así, acumulado[i] = rendimiento[0] * rendimiento[1] * ... * rendimiento[i].

Es el equivalente multiplicativo de cumsum(): perfecto para calcular rendimientos compuestos.
