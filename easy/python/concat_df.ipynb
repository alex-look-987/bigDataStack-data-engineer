# Concatenar DataFrames con índice limpio | PYTHON |fácil | Noviembre 11 de 2026

Escribe una función concatenar_limpio(lista_dfs) que reciba una lista de DataFrames (todos con las mismas columnas) y devuelva un único DataFrame que sea la concatenación de todos, con un índice numérico continuo empezando en 0.



Ejemplo (Caso 1)

df1:

| nombre	edad |
| ----------- |
| Ana	25      |
| Luis	30     |

df2:

| nombre | edad |
| ------ | ---- |
| Marta  | 28   |

Resultado:

| nombre | edad |
| ------ | ---- |
| Ana    | 25   |
| Luis   | 30   |
| Marta  | 28   |

```python
import pandas as pd
```

```python
df1 = pd.DataFrame(
    {'nombre': ['Ana', 'Luis'],
     'edad': [25, 30]}
)

df2 = pd.DataFrame(
    {'nombre': ['Marta'],
     'edad': [28]}
)

df1, df2
```

```python
def concat_dataframes(dfs: list[pd.DataFrame]):
    df = pd.concat(dfs, axis=0, ignore_index=True)

    return df
```

```python
df = concat_dataframes([df1, df2])

df
```
