```python
def fila_pascal(n: int):
    triangle = []

    for row in range(n+1):
        current = []

        for i in range(row + 1):
            if i == 0 or i == row:
                current.append(1)
            else:
                current.append(
                    triangle[row - 1][i - 1] +
                    triangle[row - 1][i]
                )
        triangle.append(current)

    return triangle[n]
```

```python
fila_pascal(4)
```

##  Optimal solution

```python
def fila_pascal(n: int) -> list:
    fila = [1]

    for i in range(1, n + 1):
        fila.append(fila[i-1] * (n - i + 1) // i)
    
    return fila
```

```python
fila_pascal(4)
```

## Explicación



Cada elemento de la fila n es el coeficiente binomial C(n, k):

1. Empezamos con [1].
2. Cada siguiente elemento se calcula multiplicando el anterior por (n - k + 1) / k.
3. Esto da exactamente C(n, 0), C(n, 1), ..., C(n, n).

Alternativa más intuitiva: construir fila a fila sumando pares de la anterior. Pero la fórmula directa es O(n) sin necesidad de construir las filas anteriores.
