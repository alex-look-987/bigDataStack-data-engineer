## Tu primer script con librerías externas


```python
# Primero crear un archivo .env con configuración
# (este archivo NO se sube a git. contiene secretos)
import os

# simular un .env escribiéndolo
with open(".env", "w") as f:
    f.write("RUTA_DATOS=data/ventas.csv\n")
    f.write("UMBRAL_VIP=500\n")
    f.write("MODO=desarrollo\n")
```

```python
# Ahora usar dotenv para cargar variables
from dotenv import load_dotenv

load_dotenv() # Cargar las variables del .env al entorno
```

```python
# Leer las variables
ruta = os.getenv("RUTA_DATOS")
umbral = int(os.getenv("UMBRAL_VIP", "100")) # 100 es el default
modo = os.getenv("MODO")

print(f"Ruta de datos: {ruta}")
print(f"Umbral VIP: {umbral}€")
print(f"Modo: {modo}")

if modo == "desarrollo":
    print("[AVISO] Ejecutando en modo desarrollo")
```
