## Comprensión de diccionario: lo mismo que ya haces, pero con claves


```python
ventas = [
    {"cliente": "Ana García", "importe": 320.00},
    {"cliente": "Luis Pérez", "importe": 890.50},
    {"cliente": "Ana García", "importe": 150.00},
]
```

```python
# La versión larga: la que ya sabes escribir
saldo = {}
for v in ventas:
    saldo[v["cliente"]] = 0.0
print(saldo)          # {'Ana García': 0.0, 'Luis Pérez': 0.0}
```

```python
# La versión corta: comprensión de diccionario
saldo = {v["cliente"]: 0.0 for v in ventas}
print(saldo)          # {'Ana García': 0.0, 'Luis Pérez': 0.0}

```

```python
# Y las dos partes pueden calcularse: clave a la izquierda, valor a la derecha
con_iva = {v["cliente"]: round(v["importe"] * 1.21, 2) for v in ventas}
print(con_iva)        # {'Ana García': 181.5, 'Luis Pérez': 1077.5}
```

## El generador: los mismos valores, sin construir la lista


```python
# con corchetes: construye la lista y luego la suma
print(sum([v['importe'] for v in ventas]))

# sin corchetes
print(sum(v['importe'] for v in ventas))

# funciona con cualquier función que recorra valores
print(max(v['importe'] for v in ventas))
print(any(v['importe'] > 500 for v in ventas))
```

## lambda: una función de usar y tirar


```python
gasto_por_cliente = {
    "Alba Requena": 2415.80,
    "Bruno Sáez": 5390.00,
    "Carla Vidal": 760.25,
}
```

```python
# con una función normal: recibe un par (cliente, gasto) y devuelve el gasto
def solo_el_gasto(par):
    return par[1]

print(sorted(gasto_por_cliente.items(), key=solo_el_gasto, reverse=True))
```

```python
# con lambda
print(sorted(gasto_por_cliente.items(), key=lambda par: par[1], reverse=True))
```

```python
# Sin key= no falla: ordena por el primer elemento del par, el nombre
print(sorted(gasto_por_cliente.items()))
# [('Alba Requena', 2415.8), ('Bruno Sáez', 5390.0), ('Carla Vidal', 760.25)]
```

```python
# Y el mismo key= sirve para quedarte con uno solo
print(max(gasto_por_cliente, key=gasto_por_cliente.get))   # Bruno Sáez
```

## map y filter: aplicar y quedarse


```python
importes = [150.0, 320.0, 85.5, 210.0, 45.0]
```

```python
# map: aplicar una operación a cada elemento
con_iva = list(map(lambda x: round(x * 1.21), importes))
print("Con IVA:", con_iva)
```

```python
# filter: quedarse solo con los que cumplen una condición
grandes = list(filter(lambda x: x > 100, importes))
print("Grandes:", grandes)
```

```python
# las dos juntas: IVA solo de los grandes
grandes_con_iva = list(map(lambda x: round(x * 1.21, 2), filter(lambda x: x > 100, importes)))
print("Grandes con IVA:", grandes_con_iva)
```

## set(): contar sin repetir


```python
pedidos = [
    {"cliente": "Ana García", "ciudad": "Sevilla"},
    {"cliente": "Luis Pérez", "ciudad": "Valencia"},
    {"cliente": "Ana García", "ciudad": "Sevilla"},
    {"cliente": "Sara Torres", "ciudad": "Valencia"},
]
```

```python
# la versión larga
vistos = []
for p in pedidos:
    if p['cliente'] not in vistos:
        vistos.append(p['cliente'])

print(len(vistos))
```

```python
# versión con un generador dentro
print(len(set(p['cliente'] for p in pedidos)))
```

```python
# un conjunto no tiene orden: para mostrarlo, se ordena
print(sorted(set(p['ciudad'] for p in pedidos)))
```

```python
# comparar dos conjuntos: quién repite y quién se ha ido
marzo = {"Ana García", "Luis Pérez", "Sara Torres"}
abril = {"Ana García", "Jorge Díaz"}

print(sorted(marzo & abril))
print(sorted(marzo - abril))
```

## defaultdict: el diccionario que no se queja


```python
from collections import defaultdict
```

```python
# La versión larga: hay que crear la clave antes de poder sumar
total = {}
for v in ventas:
    if v["cliente"] not in total:
        total[v["cliente"]] = 0.0
    total[v["cliente"]] += v["importe"]
print(total)          # {'Ana García': 470.0, 'Luis Pérez': 890.5}
```

```python
# versión corta: float() da 0.0 al estrenar cada clave
total = defaultdict(float)
for v in ventas:
    total[v["cliente"]] += v["importe"]
print(dict(total))    # {'Ana García': 470.0, 'Luis Pérez': 890.5}
```

```python
# Con list() en vez de float(): guardar todas las líneas de cada cliente
lineas = defaultdict(list)
for v in ventas:
    lineas[v["cliente"]].append(v["importe"])
print(dict(lineas))   # {'Ana García': [320.0, 150.0], 'Luis Pérez': [890.5]}
```

### Cuándo NO usarlos


```python

clientes = [
    {"nombre": "Ana García", "gasto": 7420.0, "pedidos": 14, "pais": "ES"},
]

# ILEGIBLE: tres condiciones, dos cálculos y un filtro en la misma línea
resumen = {c["nombre"]: ("VIP" if c["gasto"] / c["pedidos"] > 400 and c["pais"] == "ES" else ("revisar" if c["pedidos"] > 10 else "normal")) for c in clientes if c["pedidos"]}

# LEGIBLE: el mismo resultado, con el cálculo con nombre y las ramas separadas
resumen = {}
for c in clientes:
    if not c["pedidos"]:
        continue
    ticket_medio = c["gasto"] / c["pedidos"]
    if ticket_medio > 400 and c["pais"] == "ES":
        resumen[c["nombre"]] = "VIP"
    elif c["pedidos"] > 10:
        resumen[c["nombre"]] = "revisar"
    else:
        resumen[c["nombre"]] = "normal"

print(resumen)     # {'Ana García': 'VIP'}
```
