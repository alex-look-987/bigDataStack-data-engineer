# 02: Tu primera consulta: SELECT, FROM y WHERE

La anatomía de una consulta SQL: 

1. Seleccionar columnas, 
2. Elegir tabla 
3. Filtrar filas. 

Los tres bloques fundamentales.

Cada consulta SQL que escribas en tu vida — desde la más simple hasta la más compleja — está construida con los mismos bloques fundamentales.

Hoy vas a aprender los tres primeros: 

1. **SELECT:** qué columnas quiero
2. **FROM:** de qué tabla 
3. **WHERE:** qué filas me interesan.

Con solo estos tres ingredientes ya puedes responder preguntas reales de negocio.


## Anatomía de una consulta SQL

```sql
SELECT columna1, columna2, columna3
FROM nombre_de_tabla
WHERE condición;
```

El esqueleto de toda query SQL:

- SELECT columna1, columna2 → qué datos quieres ver (puedes poner * para todas las columnas).
- FROM nombre_de_tabla → de dónde vienen los datos.
- WHERE condición → qué filas te interesan (opcional: sin WHERE devuelve todas las filas).

Este orden de escritura es FIJO en SQL. No puedes poner WHERE antes de FROM.



- SELECT — lista las columnas que quieres ver en el resultado. Puedes poner * para "todas".
- FROM — indica de qué tabla lees los datos. Solo una tabla por ahora (luego veremos JOINs).
- WHERE — filtra filas que cumplen una condición. Es opcional: sin WHERE, obtienes TODAS las filas.
- El punto y coma (;) al final indica que la consulta terminó.

## SELECT: elegir qué columnas ver


```sql
-- Ver TODAS las columnas de la tabla clientes
SELECT * FROM clientes;

-- Ver solo nombre y ciudad (mejor práctica)
SELECT nombre, ciudad FROM clientes;

-- Puedes renombrar columnas con AS (alias)
SELECT 
    nombre AS cliente,
    email AS correo_electronico
FROM clientes;
```

- SELECT * lee TODAS las columnas. Rápido para explorar, peligroso en producción.
- SELECT nombre, ciudad — solo las columnas que necesitas. Más eficiente y predecible.
- AS crea un "alias" — un nombre temporal para la columna en el resultado. No cambia la tabla original, solo cómo se muestra. Útil para hacer resultados más legibles o para evitar nombres ambiguos en JOINs.

## FROM: elegir la tabla


FROM indica la tabla fuente. 

En nuestra base de datos tenemos: clientes, categorias, productos, pedidos y empleados. 

Por ahora trabajamos con una tabla a la vez — cuando lleguemos a JOINs combinaremos varias.

```sql
-- Todos los productos
SELECT nombre, precio FROM productos;

-- Todas las categorías
SELECT * FROM categorias;

-- Todos los pedidos
SELECT id, cliente_id, importe, fecha FROM pedidos;
```

## WHERE: filtrar filas con condiciones


WHERE es donde la magia empieza. Sin WHERE obtienes TODAS las filas. 

Con WHERE, solo las que cumplen tu condición. Es como poner un colador: solo pasan los datos que te interesan.


```sql
-- Clientes de Madrid
SELECT nombre, email
FROM clientes
WHERE ciudad = 'Madrid';

-- Productos que cuestan más de 100
SELECT nombre, precio
FROM productos
WHERE precio > 100;

-- Pedidos completados
SELECT id, importe, fecha
FROM pedidos
WHERE estado = 'completado';
```

Tres filtros distintos:

1. ciudad = 'Madrid' — igualdad con texto. El texto va SIEMPRE entre comillas simples en SQL.
2. precio > 100 — comparación numérica. Sin comillas porque 100 es un número.
3. estado = 'completado' — otra igualdad con texto.

- Cada WHERE reduce las filas devueltas. 
    - Sin WHERE, obtienes TODAS las filas de la tabla. 
    - Con WHERE, solo las que cumplen la condición.

Nota: SQL usa un solo = para igualdad (no == como Python). Es una fuente común de confusión para programadores.

## Operadores de comparación


- = → igual a (OJO: es un solo igual, no doble como en Python)
- != o <> → distinto de
- '>' → mayor que
- < → menor que
- '>'= → mayor o igual
- <= → menor o igual

## Combinar condiciones: AND, OR, NOT


Puedes combinar múltiples condiciones:

1. AND (todas deben cumplirse)
2. OR (al menos una) 
3. NOT (negar). 

- Cuando mezclas AND y OR, usa paréntesis para que la precedencia sea clara.


```sql
-- Clientes VIP de Madrid
SELECT nombre, ciudad
FROM clientes
WHERE es_vip = TRUE AND ciudad = 'Madrid';

-- Clientes de Madrid o Barcelona
SELECT nombre, ciudad
FROM clientes
WHERE ciudad = 'Madrid' OR ciudad = 'Barcelona';

-- Clientes que NO son VIP
SELECT nombre, ciudad
FROM clientes
WHERE NOT es_vip;

-- Cuidado con la precedencia: usa paréntesis
SELECT nombre, ciudad, es_vip
FROM clientes
WHERE (ciudad = 'Madrid' OR ciudad = 'Sevilla') AND es_vip = TRUE;
``` 

Tres variantes de condiciones compuestas:

- AND: AMBAS condiciones deben ser TRUE. El resultado es la intersección (VIP Y de Madrid).
- OR: al menos UNA debe ser TRUE. El resultado es la unión (Madrid O Barcelona).
- Paréntesis: fuerzan el orden de evaluación. Sin ellos, AND tiene mayor precedencia que OR, lo que puede dar resultados inesperados.

**Regla:** cuando mezcles AND y OR en la misma cláusula WHERE, usa SIEMPRE paréntesis para hacer explícita tu intención. Es más legible y evita bugs sutiles.

- AND tiene mayor precedencia que OR. 
- Sin paréntesis, "ciudad = Madrid OR ciudad = Sevilla AND es_vip = TRUE" se interpreta como "Madrid O (Sevilla Y VIP)" — probablemente no es lo que querías. 
- Cuando mezcles AND y OR, usa SIEMPRE paréntesis. 

Y no te lo creas porque lo diga aquí: copia las dos consultas del ejemplo de arriba, la de con paréntesis y la de sin, y compara cuántas filas te dice el editor en cada caso. La diferencia es de varias veces, por dos caracteres.



## Las tablas con las que vas a trabajar


```sql
-- Ver qué tablas hay
SHOW TABLES;

-- Ver las columnas de cada table
DESCRIBE clientes;
DESCRIBE pedidos;
DESCRIBE productos;
```
