## [03] Devoluciones de unos días concretos

Atención al cliente revisa las devoluciones de los tres primeros días de febrero de 2024, del 1 al 3 incluido. 

Muestra, en este orden: id, importe y fecha.


| id    | importe | fecha      |
| ----- | ------- | ---------- |
| 17808 | 124.91  | 2024-02-01 |
| 10558 | 12.88   | 2024-02-01 |
| 12014 | 67.25   | 2024-02-03 |
| 15862 | 18.02   | 2024-02-03 |
| 6487  | 11.76   | 2024-02-01 |
| 11192 | 169.83  | 2024-02-02 |

```sql
SELECT id, importe, fecha
FROM pedidos
WHERE estado = 'devuelto'
    AND fecha >= '2024-02-01'
    AND fecha < '2024-02-04';

```
