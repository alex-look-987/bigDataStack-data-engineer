# I/O Files

### open() and with: the door to the files system

```python
# Escribir un archivo de texto
with open("datos_ejemplo.txt", 'w', encoding='utf-8') as f:
    f.write('producto,cantidad,precio\n')
    f.write("Laptop,5,1299.99\n")
    f.write("Monitor,12,349.50\n")
    f.write("Teclado,30,79.99\n")
```

```python
# Leer el archivo completo
with open("datos_ejemplo.txt", "r", encoding="utf-8") as f:
    contenido = f.read()
print(contenido)
```

```python
# Leer línea por línea (mejor para archivos grandes)
with open("datos_ejemplo.txt", "r", encoding="utf-8") as f:
    print(f)
    for linea in f:
        print(linea.strip()) # strip() quitan el \n del final
    
```

```python
# Contarlíneas sin recorrerlas una a una, devolver la lista entera de golpe
with open("datos_ejemplo.txt", "r", encoding="utf-8") as f:
    lineas = f.readlines() # lista con una entrada por línea
    print(f"El fichero tiene {len(lineas)} líneas")
    print(lineas)
```

### Reading CSVs files with csv module

```python
import csv

# creamos un CSV de ejemplo
with open("ventas.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(['fecha', 'producto', "cantidad", "precio"])
    writer.writerow(["2024-01-15", "Laptop Pro", 2, 1299.99])
    writer.writerow(["2024-01-15", "Monitor 4K", 1, 449.00])
    writer.writerow(["2024-01-16", "Teclado mecánico", 5, 89.99])
    writer.writerow(["2024-01-16", "Ratón ergonómico", 3, 59.99])
```

```python
# Leer el CSV como lista de diccionarios (DictReader)
with open("ventas.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    ventas = list(reader)
```

```python
# Ventas es una lista de diccionarios
for venta in ventas:
    total = int(venta['cantidad']) * float(venta['precio'])
    print(f"{venta['fecha']} | {venta['producto']:20s} | {total:.2f}€")

```

```python
# Convertir una lista de diccionarios a CSV
filas = [
    {"producto": "Laptop", "total": "3899.97"},
    {"producto": "Monitor", "total": "1995.00"},
]

with open("salida.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=['producto', 'total'])
    writer.writeheader() # escribe las filas de nombres
    writer.writerows(filas) # escribe todas las filas de golpe
```

```python
# utf-8-sig se come la marca invisible; delimiter le dice cuál es el separador de verdad
with open("ventas_excel.csv", "w", encoding="utf-8-sig", newline="") as f:
    f.write("producto;cantidad;precio\n")
    f.write("Laptop;3;1299,99\n")
    f.write("Monitor;2;449,00\n")

with open("ventas_excel.csv", "r", encoding="utf-8-sig", newline="") as f:
    lector = csv.DictReader(f, delimiter=";")

    for fila in lector:
        # el decimal de coma a punto antes de convertirlo a número
        precio = float(fila['precio'].replace(",", "."))
        print(f"{fila['producto']}: {precio:.2f}€")
```

### JSON: el formato de las APIs

```python
import json

# datos que queremos guardar
resumen_diario = {
    "fecha": "2024-01-15",
    "total_ventas": 3057.96,
    "num_transacciones": 4,
    "productos_top": ["Laptop Pro", "Monitor 4K"],
    "meta_alcanzada": True,
}
```

```python
# Escribir JSON (serializar)
with open("resume.json", "w", encoding="utf-8") as f:
    json.dump(resumen_diario, f, indent=2, ensure_ascii=False)
```

```python
# Leer JSON (deserializar)

with open("resume.json", "r", encoding="utf-8") as f:
    datos = json.load(f)

print(f"Fecha: {datos['fecha']}")
print(f"Total: {datos['total_ventas']}€")
print(f"Top: {datos['productos_top']}")
```

### Rutas de archivos: el dolor silencioso

```python
from pathlib import Path

# pathlib funciona igual en windows y mac
ruta_datos = Path("data") / "raw" / "ventas.csv"
print(ruta_datos) # data/raw/ventas.csv (Mac) o data\raw\ventas.csv (Win)
```

```python
# Crear directorios si no existen
ruta_datos.parent.mkdir(parents=True, exist_ok=True)
```

```python
# Verificar si existe 
if ruta_datos.exists():
    print(f"Tamaño: {ruta_datos.stat().st_size} bytes")
else:
    print("El archivo no existe aún")
```

```python
# Listar archivos de un directorio
for archivo in Path('data').glob("*.csv"):
    print(f"Encontrado: {archivo.name}")
```
