# 04: Imágenes y Dockerfiles: la receta de tu entorno


1. Estás en ~. Entra en la carpeta del proyecto (ya tiene el Dockerfile).
    - cd mi-etl
2. Construir la imagen una vez
    - docker build -t mi-etl:1.2.0 .
3. Añadir un tag de fecha (sin reconstruir)
    - docker tag mi-etl:1.2.0 mi-etl:___
4. Añadir el tag "latest" (sin reconstruir)
    - docker tag mi-etl:1.2.0 mi-etl:___
5. Comprueba que los tres tags apuntan al MISMO id:
    - docker images
