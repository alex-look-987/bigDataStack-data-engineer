## [01] Tu primer SELECT: SQL como calculadora


SQL no necesita tablas para funcionar. 

Puedes usar SELECT como una calculadora o para obtener información del sistema. 

Escribe tres consultas: 

1. Una que sume 1+1
2. otra que devuelva el texto "Hola mundo" (usando || para concatenar)
3. Otra que use typeof() para averiguar de qué tipo es CURRENT_DATE. 



```sql
-- 1. Usa SELECT para calcular 1 + 1
SELECT 1 + 1 AS resultado;

-- 2. Usa SELECT para devolver 'Hola' || ' ' || 'mundo'
SELECT 'Hola' || ' ' || 'mundo' AS saludo;

-- 3. Usa typeof() sobre CURRENT_DATE para ver su tipo
SELECT typeof(CURRENT_DATE) AS tipo_fecha;
```

## [02] Explorar las tablas del e-commerce


Acabas de llegar a un proyecto nuevo. Lo primero: explorar. 

1. Usa SHOW TABLES para ver qué tablas hay, 
2. luego DESCRIBE pedidos para ver su estructura, 
3. finalmente cuenta cuántas filas tiene la tabla pedidos.

```sql
-- 1. Ver todas las tablas disponibles
SHOW TABLES;

-- 2. Ver la estructura de la tabla pedidos
--    Escribe aquí:  DESCRIBE pedidos;
DESCRIBE pedidos;

-- 3. Contar cuántas filas tiene pedidos
SELECT COUNT(*) AS total_pedidos FROM pedidos;
```

## [03] De pseudocódigo imperativo a SQL declarativo



Abajo tienes un pseudocódigo imperativo que recorre una lista de empleados paso a paso para encontrar los que ganan más de 95000. 

1. Tu tarea: 
   - escribe la consulta SQL equivalente (declarativa). 
2. Recuerda: 
    - en SQL dices QUÉ quieres, no CÓMO buscarlo.

```sql
-- PSEUDOCÓDIGO IMPERATIVO (lo que NO queremos hacer):
-- resultado = []
-- PARA CADA fila EN empleados:
--     SI fila.salario > 95000:
--         resultado.añadir(fila.nombre, fila.salario)
-- DEVOLVER resultado

-- Escribe el SQL DECLARATIVO equivalente:
SELECT nombre, salario
FROM empleados
WHERE salario > 95000
```
