# 06: Volúmenes y redes: persistir datos y comunicar contenedores


## [01] PostgreSQL con persistencia

1. Añade un volumen con nombre para que los datos de PostgreSQL sobrevivan a docker compose down.
2. Estás en la carpeta ventas/, con el docker-compose.yml sin persistencia y con huecos (___). 
    - Rellénalos en el panel de arriba —el montaje del servicio db y el volumen ventas_data a nivel raíz— y valida con `docker compose config --volumes`.

```yml
services:
  db:
    image: postgres:16
    environment:
      POSTGRES_PASSWORD: prod2024
      POSTGRES_DB: ventas
    ports:
      - "5432:5432"
    volumes:
      - ventas_data:/var/lib/postgresql/data

volumes:
  ventas_data:
```

```bash
# Comprobar el volumen
docker compose config --volume
```

## [02] Bind mount para desarrollo

1. Sincroniza tu código local con el contenedor: editas en VS Code y el contenedor lo ve al instante. 
    - El punto (./) es la clave.
2. Estás en la carpeta etl-dev/. 
    - El docker-compose.yml ya existe con huecos (___) para los dos bind mounts: ./src y ./output. 
    - Rellénalos en el panel de arriba y valida con `docker compose config`.

```yml
services:
  etl:
    build: .
    volumes:
      - ./src:/app/src
      - ./output:/app/output
    environment:
      ENV: development
```

```bash
# Validar
docker compose config --services
```


## [03] Limpieza segura de Docker

1. Disco lleno. 
    - Hay un contenedor parado con su imagen, y un volumen con nombre (ventas_data) con datos que NO puedes perder.
    -  Limpia en orden pero deja ventas_data intacto.
2. En el panel de la derecha verás que ventas_data sigue ahí al terminar. 
    - La clave: docker volume prune (sin -a) solo borra volúmenes anónimos; los que tienen nombre están a salvo.

```bash
# Comandos a probar

# 1. Ver cuánto hay por limpiar
docker system df

# 2. Listar volúmenes (ventas_data tiene datos que NO puedes perder)
docker volume ls

# 3. Borrar los contenedores parados (retienen las imágenes)
docker container prune -f

# 4. Borrar las imágenes que ya no usa nadie
docker image prune -a -f

# OJO: NO uses 'docker volume prune -a' — ese borraría ventas_data.
# Comprueba que sigue:
docker volume ls
```
