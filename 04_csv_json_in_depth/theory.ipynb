## JSON Lines: El formato para datos grandes

JSON clásico tiene un problema con archivos grandes: necesitas cargar TODO el archivo en memoria para parsearlo (porque es un único array/objeto).

- Si tienes 2 GB de datos, necesitas 2 GB de RAM solo para leerlo. 
- La solución: JSON Lines (JSONL). 
- Un objeto JSON por línea, sin array contenedor. 
- Puedes leer línea a línea sin cargar todo en memoria.


```python
import json
```

```python
# --- JSON clásico (todo en un array) ---
# Problema: necesitas cargar TODO en memoria
json_clasico = '[{"id":1,"nombre":"Ana"},{"id":2,"nombre":"Pedro"}]'
todos = json.loads(json_clasico)  # Carga TODO de golpe
```

```python
# --- JSON Lines (un objeto por línea) ---
# Ventaja: puedes leer línea a línea
jsonl_contenido = """{"id":1,"nombre":"Ana"}
{"id":2,"nombre":"Pedro"}
{"id":3,"nombre":"Lucía"}
"""
```

```python
# Leer lína a línea (eficiencia en memoria)
for linea in jsonl_contenido.strip().split("\n"):
    registro = json.loads(linea)
    print(f"Procesando: {registro['nombre']}")
```

```python
# En un archivo real:
with open("errores_dia.jsonl", encoding='utf-8') as f:
    data = json.load(f)

for d in data:
    print(f"Fila {d['fila']} - Error: {d['error']}")
    print(f"Datos afectados: {d['datos']}\n")
```

## Leer CSV con problemas reales en Python

```python
import csv
```

```python
# Escribir un CSV problemático para practicar
with open("ventas_problematicas.csv", 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)

    writer.writerow(['producto', 'descripcion', 'precio', 'stock'])
    writer.writerow(['Camiseta', 'Algodón, talla M', '29.99', '150'])
    writer.writerow(['Pantalón', 'Vaquero "slim fit"', '59.90', ''])
    writer.writerow(['Chaqueta', 'Impermeable\nlarga', '120.00', '45'])
```

```python
# Leer manejando problemas
with open('ventas_problematicas.csv', 'r', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)

    for fila in reader:
        # Convertir precio a float (viene como texto)
        precio = float(fila['precio'])

        # Manejar stock vacío como 0
        stock = int(fila['stock'])  if fila['stock'] else 0
        print(f"{fila['producto']:12} | {precio:>7.2f} | stock: {stock}")
```
