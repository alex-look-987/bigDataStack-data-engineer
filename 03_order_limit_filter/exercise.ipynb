# 03: Ordenar, limitar y filtrar: controlando los resultados

ORDER BY, LIMIT, DISTINCT, LIKE, BETWEEN e IN — las herramientas para refinar tus consultas con precisión.

## [01] Los 5 pedidos más grandes



El CFO quiere ver los 5 pedidos de mayor importe. Muestra, en este orden: id, importe y fecha. Ordena del mayor importe al menor, desempatando por id.

Resultado esperado

| id   | importe | fecha      |
| ---- | ------- | ---------- |
| 298  | 319.2   | 2023-01-14 |
| 1545 | 319.2   | 2023-08-13 |
| 2792 | 319.2   | 2023-04-13 |
| 4039 | 319.2   | 2023-06-29 |
| 5286 | 319.2   | 2024-06-02 |



```sql
-- Top 5 pedidos por importe (desempatada por id)
SELECT id, importe, fecha
FROM pedidos
ORDER BY importe DESC NULLS LAST, id
LIMIT 5
```

## [02] Estados de pedido



Atención al cliente quiere saber qué estados de pedido existen en la base de datos. Muestra los estados únicos, ordenados alfabéticamente.

| estado     |
| ---------- |
| cancelado  |
| completado |
| devuelto   |
| NULL       |

```sql
-- Estados únicos de pedidos oerdenados alfabéticamente

SELECT DISTINCT estado
FROM pedidos
ORDER BY estado;
```

## [03] Productos en rango de precio

El equipo de diseño necesita los productos entre 25€ y 80€ para la sección "Regalos accesibles". 

Muestra, en este orden: 

1. nombre y precio, 
2. ordenados por precio 
3. desempatando por nombre. 

El ORDER BY y el LIMIT 10 del final ya están puestos: 
- la comprobación automática solo mira las 10 primeras filas
- Sin un orden fijo no siempre son las mismas.

Resultado esperado

| nombre                | precio |
| --------------------- | ------ |
| Manual de Redacción   | 28.93  |
| Lámpara Nórdica 9     | 33.04  |
| Manual de Cocina      | 33.31  |
| Manual de Fotografía  | 33.69  |
| Lámpara Nórdica 12    | 33.82  |
| Manual de Diseño      | 34.07  |
| Batidora Turbo 5      | 34.47  |
| Batidora Turbo 12     | 35.81  |
| Batidora Clásica 3    | 37.16  |
| Manual de Estadística | 37.32  |

```sql
-- Productos entre 25 y 80 euros, ordenados por precio
SELECT nombre, precio
FROM productos
WHERE precio BETWEEN 25 AND 80
ORDER BY precio, nombre
LIMIT 10;
```

## [04] Pedidos problemáticos

Atención al cliente necesita los pedidos en estado "cancelado" o "devuelto". 

Muestra, en este orden: 

1. id, cliente_id, importe y estado. 
2. Ordena por id. 

El ORDER BY y el LIMIT 10 del final ya están puestos: 
- la comprobación automática solo mira las 10 primeras filas
- Sin un orden fijo no siempre son las mismas.

Resultado esperado

| id  | cliente_id | importe | estado    |
| --- | ---------- | ------- | --------- |
| 3   | 821        | 44.01   | devuelto  |
| 6   | 1099       | 138.26  | devuelto  |
| 9   | 1955       | 11.93   | cancelado |
| 12  | 2258       | 50.41   | cancelado |
| 15  | 1592       | 154.07  | cancelado |
| 18  | 409        | 14.22   | cancelado |
| 21  | 2816       | 57.53   | cancelado |
| 24  | 1543       | 171.28  | cancelado |
| 47  | 244        | 163.19  | devuelto  |
| 50  | 2465       | 15.59   | devuelto  |

```sql
-- Pedidos cancelados o devueltos, ordenados por id

SELECT id, cliente_id, importe, estado
FROM pedidos
WHERE estado IN ('cancelado', 'devuelto')
ORDER BY id
LIMIT 10;
```
